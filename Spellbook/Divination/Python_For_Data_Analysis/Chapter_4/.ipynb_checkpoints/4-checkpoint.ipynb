# 4.0 Introduction to Numpy

```python
import numpy as np
```

```python
array_spell = np.arange(1_000_000)
```

```python
my_list = list(range(1_000_000))
```

```python
%timeit array_mult_2_spell = array_spell * 2
```

```python
%timeit list_mult_2_spell = [x*2 for x in my_list]
```
