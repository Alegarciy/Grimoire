### 4.1.1.4 Indexing and Slicing

```python
arr = np.arange(10)
arr[5]
```

```python
arr[5:8]
```

```python
arr[5:8] = 12
arr
```

```python
# Indexing with slicing

# Syntathic Sugar
# np.tile(row, (number_of_rows_repeats, number_of_column_repeats))
# array = np.tile(arange(10), (2,1))

array = np.array([np.arange(10),np.arange(10),np.arange(10)])
array
```

```python
# Axis 0 slicing
array[:2]
```

```python
# Axis 0 and Axis 1 slicing
array[:2,:5]
```

```python
# Axis 1 slicing
array[:,:5]
```

```python
# Boolean Indexing

names = np.array(["Bob","Joe","Will","Bob","Will","Joe","Joe"])
names == "Bob"
```

```python
data = np.array([
[4,7], # Bob
[0,2],
[-5,6],
[0,0], # Bob
[1,2],
[-12,-4],
[3,4]
])

data[names == "Bob"]
```

```python
# Boolean Negation
~(names == "Bob")
```
